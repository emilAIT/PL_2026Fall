```python
def poshitay(a, b, c, d, e, f, g, h, k):
    total = a*b + c + d + e + f + a*b*g + h + k 

    total = total + total * 15/100 # (1 + 0.15) * total  -> 1.15 * total

    price = total / b

    return price
```

```python
d1, m1, y1 = 5, 9, 2008
d2, m2, y2 = 7, 9, 2008
res = None

def sravni(d1, m1, y1, d2, m2, y2):
    if y1 > y2: 
        res = 1
    elif y2 > y1:
        res = 2
    elif m1 > m2:
        res = 1
    elif m2 > m1:
        res = 2
    elif d1 > d2:
        res = 1
    elif d2 > d1:
        res = 2
    else:
        res = 1

    if res == 1:
        return d1, m1, y1
    else:
        return d2, m2, y2


def mladshiy(d1, m1, y1, d2, m2, y2, d3, m3, y3):
    a, b, c = sravni(d1, m1, y1, d2, m2, y2)
    x, y, z = sravni(a, b, c, d3, m3, y3)
    return x, y, z
```
