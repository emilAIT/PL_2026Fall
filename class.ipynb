```python
students = [] # student names
grades = [] # quizes of each student
midterms = [] # midterms (two midterms)
final = [] # final score 

def krassiviy(arr1, arr2):
    s = ''
    for i, j in zip(arr1, arr2):
        s += f'{i} = {j}\n'
    return s


def add_student(name):
    students.append(name)
    grades.append([])
    midterms.append([])
    final.append(0)

    return krassiviy(students, grades)


def remove_student(name):
    for i, j in enumerate(students):
        if j == name:
            students.pop(i)
            grades.pop(i)
            midterms.pop(i)
            final.pop(i)
    return krassiviy(students, grades)

def get_grade(name):
    for i, j in enumerate(students):
        if j == name:
            return (name, grades[i])


def update_grade(name, index, score):
    for i, j in enumerate(students):
        if j == name:
            if index < len(grades[i]):
                grades[i][index] = score
            else:
                return f'{name} has less grades than index = {index}, {grades[i]}'
    return krassiviy(students, grades)

def add_grade(name, grade):
    if name in students:
        for i, v in enumerate(students):
            if v == name:
                grades[i].append(grade)
        return krassiviy(students, grades)
    else:
        return f'{name} does not exist in {students}' 

def add_midterm(name, grade):
    if name in students:
        for i, v in enumerate(students):
            if v == name:
                midterms[i].append(grade)
    else:
        print(f'{name} does not exists in {students}')

def add_final(name, grade):
    if name in students:
        for i, v in enumerate(students):
            if v == name:
                final[i] = grade
    else:
        print(f'{name} does not exists in {students}')

def student_average():
    s = ''
    for name, grade in zip(students, grades):
        x = sorted(grade)
        avg = sum(x[-13:]) // 13
        s += f'{name} : {avg} \n'
    return s
        

```

```python
RATES = {  # курс к сому: покупка пунктом / продажа пунктом
    "USD": {"buy": 89.0, "sell": 89.5}, 
    "EUR": {"buy": 96.0, "sell": 96.8},
    "RUB": {"buy": 0.98, "sell": 1.02}, 
    "KZT": {"buy": 0.19, "sell": 0.20},
    "CNY": {"buy": 12.3, "sell": 12.5},
}


history = []


def transaction_cost(currency, amount, op, client_id):
    if op not in ['buy', 'sell'] or amount <= 0 or currency not in RATES:
        return -1
    history.append((client_id, currency, amount, op))
    return amount * RATES[currency][op]

def filter_deals(client_id=None, op=None):
    result = []
    if not client_id and not op:
        return history
    elif not op:
        for i in history:
            if i[0] == client_id:
                result.append(i)
    elif not client_id:
        for i in history:
            if i[-1] == op:
                result.append(i)
    return result


def currency_summary(currency):
    bought = 0
    sold = 0
    sum_bought = 0
    sum_sold = 0

    for _, currency, amount, op in history:
        if op == 'buy':
            bought += amount
            sum_bought += amount * RATES[currency]['buy']
        else:
            sold += amount
            sum_sold += amount * RATES[currency]['sell']

    return {'bought':bought, 'sold': sold, 'avg_buy': sum_bought/bought, 'avg_sell': sum_sold/sold}        


def gross_profit():
    total = 0
    for currency in RATES:
        d = currency_summary(currency)
        total += d['sold'] * (d['avg_sell'] - d['avg_buy'])
    return total


def top_clients(min_usd):
    clients = {}
    for client, currency, amount, op in history:
        if client not in clients:
            clients[client] = 0
        
        clients[client] += RATES[currency][op] * amount

    result = []
    for client, total in clients.items():
        if total/RATES['USD']['sell'] >= min_usd:
            result.append((amount, client))
    return sorted(result)[::-1]


```

```python
import random

def encode_exam_sheet(text: str, a, b, c, d) -> str:
    # 1. Convert characters to 20-bit binary strings
    bitstream = "".join(f"{ord(char):0{a}b}" for char in text)
    
    # 2. Pad bitstream with trailing zeros so length is a multiple of 26
    remainder = len(bitstream) % b
    if remainder != 0:
        bitstream += "0" * (b - remainder)
        
    # 3. Split into 26-bit chunks and convert to 10-digit zero-padded decimal strings
    decimal_chunks = []
    for i in range(0, len(bitstream), b):
        chunk_26 = bitstream[i:i+b]
        number = int(chunk_26, 2)
        decimal_chunks.append(f"{number:010d}")
        
    # 4. Append 11 random 10-digit numbers as trailing noise
    # (Generating valid numbers that fit within 26-bit ceiling: max 67108863)
    for _ in range(d):
        fake_num = random.randint(0, (1 << 26) - 1)
        decimal_chunks.append(f"{fake_num:010d}")
        
    # Continuous digit stream output
    return "".join(decimal_chunks)


def decode_exam_sheet(digit_stream: str, a, b, c, d: int) -> str:
    # 1. Parse 10-digit decimal blocks
    chunks = [digit_stream[i:i+c] for i in range(0, len(digit_stream), c)]
    
    # 2. Strip off the 11 trailing garbage chunks
    if d > 0:
        chunks = chunks[:-d]
        
    # 3. Convert decimal blocks back to 26-bit binary strings
    bitstream = "".join(f"{int(num):0{b}b}" for num in chunks)
    
    # 4. Extract 20-bit character chunks
    chars = []
    for i in range(0, len(bitstream) - (len(bitstream) % a), a):
        chunk_20 = bitstream[i:i+a]
        code_point = int(chunk_20, 2)
        if code_point != 0:  # Ignore trailing padding zeros
            chars.append(chr(code_point))
            
    return "".join(chars)

encoded_stream = encode_exam_sheet(s, 20, 26, 10, 13)
decoded_text = decode_exam_sheet(encoded_stream, 20, 26, 10, 13)

print(f"Encoded Output ({len(encoded_stream)} digits):\n{encoded_stream}\n")
print(f"Decoded Output:\n{decoded_text}")
```

```python
class aitarray:
    def __init__(self, op = 0):
        self.data = []
        self.operatsi = op

    def dobavit_zadi(self, val):
        self.data = self.data + [val]
        self.operatsi += 1

    def udali(self, val):
        inx = -1
        for i, j in enumerate(self.data):
            if j == val:
                inx = i
                break
        if inx != -1:
            self.data = self.data[:inx] + self.data[inx+1:]
        self.operatsi += 1

    def udali_vse(self, val):
        data = []
        for i in self.data:
            if i != val:
                data += [i]
        self.data = data
        self.operatsi += 1

    def udali_vse2(self, val):
        while val in self.data:
            self.udali(val)


    def dobavit_po_indeksu(self, index, val):
        if index < len(self.data):
            self.data = self.data[:index] + [val] + self.data[index:]
        else:
            print('out of range index')
        self.operatsi += 1

    def uberi(self, index = -1):
        if self.data:
            if index == -1:
                index = len(self.data) - 1
            val = self.data[index]
            self.data = self.data[:index] + self.data[index+1:]
            return val
        else:
            print('index out of range or empty array')
        self.operatsi += 1

    def total_operatsi(self):
        return self.operatsi

    def pokaji(self):
        print('Massiv: ', self.data)
    
```

```python
class samalot:
    def __init__(self, imya_p, rows, cols, tsena):
        self.imya_reysa = imya_p
        self.kolichestvo = len(rows) * len(cols)
        self.sidenie = {}
        self.price = {}
        for i in rows:
            for j in cols:
                self.dobav_sidenie(f'{i}{j}')
                self.price[f'{i}{j}'] = tsena

    def pomenyat_tsenu(self, mesta, noviy_tsena):
        if mesta in self.price:
            self.price[mesta] = noviy_tsena

    def total_oborot(self):
        count = 0
        for i in self.sidenie:
            if self.sidenie[i] != None:
                count += self.price[i]
        return count
        
    def dobav_sidenie(self, nomer_sidenie):
        self.sidenie[nomer_sidenie] = None
    
    def add_passenger(self, pasajir, nomer_sidenie):
        if nomer_sidenie in self.sidenie and self.sidenie[nomer_sidenie] == None:
            self.sidenie[nomer_sidenie] = pasajir
            self.kolichestvo -= 1
        else:
            print('izvinite ne letite')

    def cancel(self, pasajir, nomer_sidenie):
        if nomer_sidenie in self.sidenie and self.sidenie[nomer_sidenie] == pasajir:
            self.sidenie[nomer_sidenie] = None
            self.kolichestvo += 1
        else:
            print('ne pravilno veli dannie')

    def pokaji(self):
        print('imya reysa: ', self.imya_reysa)
        for i in self.sidenie:
            if self.sidenie[i] != None:
                print(i, self.sidenie[i], self.price[i])
        print('Kolchesvto svobodnih: ', self.kolichestvo)
        print('Oborot: ', self.total_oborot())
        print()
    
    def find_person(self, imya):
        return imya in self.sidenie.values()
        





```
